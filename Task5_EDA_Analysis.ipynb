# Task 5: Exploratory Data Analysis (EDA)
## Titanic-style passenger dataset

**Objective:** Extract insights using visual and statistical exploration.  
**Tools:** Python, Pandas, Matplotlib, Seaborn.  
**Important dataset note:** This project includes a *simulated Titanic-style dataset* generated for this assignment, not the original Titanic/Kaggle dataset. Replace `data/titanic_style_sample.csv` with the official dataset if your instructor requires the original source.

### Questions explored
- What does the dataset look like?
- Which columns contain missing values?
- How are age and fare distributed?
- How do survival outcomes differ by sex and passenger class?
- What numeric relationships appear in a correlation heatmap?

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

DATA_PATH = Path("data/titanic_style_sample.csv")
df = pd.read_csv(DATA_PATH)
df.head()
```

## 1. Dataset overview
`head()`, `shape`, `info()` and `describe()` help understand rows, columns, data types and numeric distributions.

```python
print("Rows and columns:", df.shape)
display(df.head())
df.info()
display(df.describe(include="all").T)
```

## 2. Missing values and duplicates
Missing values are counted per column. For a real analysis, decide whether to impute, remove, or retain missing values based on context.

```python
missing = df.isna().sum().sort_values(ascending=False)
display(missing[missing > 0])
print("Duplicate rows:", df.duplicated().sum())
```

## 3. Value counts
`value_counts()` summarizes categorical columns and outcome counts.

```python
for col in ["Survived", "Pclass", "Sex", "Embarked"]:
    print(f"\\n{col}")
    display(df[col].value_counts(dropna=False))
```

## 4. Univariate analysis: histograms and boxplots

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
sns.histplot(data=df, x="Age", bins=25, kde=True, ax=axes[0])
axes[0].set_title("Age Distribution")
sns.histplot(data=df, x="Fare", bins=30, kde=True, ax=axes[1])
axes[1].set_title("Fare Distribution")
plt.tight_layout()
plt.show()

plt.figure(figsize=(7, 4))
sns.boxplot(data=df, x="Survived", y="Age")
plt.title("Age by Survival Outcome")
plt.show()
```

**Observation:** Fare commonly shows right-skew in passenger datasets: a small number of passengers pay much higher fares. Boxplots help identify spread and possible outliers. Confirm the actual pattern from the charts when you run the notebook.

## 5. Bivariate analysis: survival by sex and passenger class

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
sns.barplot(data=df, x="Sex", y="Survived", errorbar=None, ax=axes[0])
axes[0].set_title("Survival Rate by Sex")
axes[0].set_ylabel("Mean survival (0–1)")
sns.barplot(data=df, x="Pclass", y="Survived", errorbar=None, ax=axes[1])
axes[1].set_title("Survival Rate by Passenger Class")
axes[1].set_ylabel("Mean survival (0–1)")
plt.tight_layout()
plt.show()

display(df.groupby("Sex")["Survived"].mean().rename("survival_rate"))
display(df.groupby("Pclass")["Survived"].mean().rename("survival_rate"))
```

**Observation:** Compare bar heights to see how survival rate varies across groups. These are descriptive associations, not proof that a single factor caused an outcome.

## 6. Multivariate analysis: pairplot and correlation heatmap

```python
numeric_cols = ["Survived", "Pclass", "Age", "SibSp", "Parch", "Fare"]
sns.pairplot(df[numeric_cols].dropna().sample(min(250, len(df.dropna())), random_state=42))
plt.show()

plt.figure(figsize=(9, 6))
sns.heatmap(df[numeric_cols].corr(), annot=True, cmap="coolwarm", fmt=".2f", center=0)
plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()
```

**Observation:** Correlation measures linear association between numeric columns. A correlation does not imply causation. Pairplots help inspect distributions and pairwise relationships; heatmaps provide a compact summary of correlations.

## 7. Outlier and skewness checks

```python
print("Fare skewness:", df["Fare"].skew())
q1, q3 = df["Fare"].quantile([0.25, 0.75])
iqr = q3 - q1
lower, upper = q1 - 1.5 * iqr, q3 + 1.5 * iqr
fare_outliers = df[(df["Fare"] < lower) | (df["Fare"] > upper)]
print("Potential fare outliers:", len(fare_outliers))
display(fare_outliers.head())
```

## 8. Summary of findings
Use the calculated results and plots to describe key patterns, data quality issues, and limitations. This project uses simulated data, so do not present its rates as historical facts about the actual Titanic passengers.

```python
print(f"Overall survival rate: {df['Survived'].mean():.1%}")
print("\\nSurvival rate by sex:")
display(df.groupby("Sex")["Survived"].mean())
print("\\nSurvival rate by passenger class:")
display(df.groupby("Pclass")["Survived"].mean())
print("\\nMissing values:")
display(df.isna().sum())
```

## Interview questions — short answers
1. **What is EDA and why is it important?** Exploratory Data Analysis uses summaries and visualizations to understand data quality, patterns, relationships and anomalies before deeper analysis.
2. **Which plots check correlation?** Correlation heatmaps and scatterplots; pairplots can inspect several numeric relationships.
3. **How do you handle skewed data?** Inspect the distribution, verify outliers, consider log/other transformations when appropriate, and report the method.
4. **How do you detect multicollinearity?** Check correlation among predictors and use Variance Inflation Factor (VIF) where appropriate.
5. **Univariate, bivariate and multivariate analysis?** One variable, two variables, and three or more variables respectively.
6. **Heatmap vs pairplot?** Heatmap compactly encodes a matrix of values such as correlations; pairplot shows distributions and pairwise scatterplots.
7. **How do you summarize insights?** State the pattern, give supporting numbers/plots, explain limitations, and connect it to the question being investigated.
